# Library

The same functions as in [experimental_result_analysis.ipynb](./experimental_result_analysis.ipynb)

Extract the ground truth and result

```python
def extract_result(row, ground_truth, win_label="CHATGPT", lose_label="VICUNA13B", tie_label="TIE"):
    first_score, second_score = row['score']

    row['result'] = win_label if round(first_score, 2) > round(second_score, 2) else lose_label if round(first_score,
                                                                                                         2) < round(
        second_score, 2) else tie_label
    row['ground_truth'] = ground_truth[row.name - 1]
    row['result_hit'] = row['result'] == row['ground_truth']

    return row
```

Reverse score

```python
def reverse_row(row):
    first_score, second_score = row['score']
    is_reversed = row['is_reversed']

    if is_reversed:
        row['score'] = [second_score, first_score]
        row['text_gpt'], row['text_vicuna'] = row['text_vicuna'], row['text_gpt']

    return row
```

# Vicuna dataset sampling for CJ experiments

This section performs the sampling of the Vicuna dataset for CJ experiments.  
The workflow:
1. Load review results and filter out ties.
2. Load questions and model answers.
3. Randomly sample one question per category.
4. Save the sampled subset for all models.

```python
def save_items(df, id_field, id_list, save_path):
    """Filter dataframe by IDs and save as JSONL."""
    df[df[id_field].isin(id_list)].to_json(save_path, orient="records", lines=True)
```

## Load review results and questions

We read the human review file, attach the ground truth to the questions.


```python
VICUNA_DATASET_PATH = "../datasets/vicuna"
```

```python
with open(f'{VICUNA_DATASET_PATH}/ground-truth/cj_ground_truth.txt', 'r', encoding='utf-8') as f:
    text_lines = f.read().splitlines()

question_df = pd.read_json(f'{VICUNA_DATASET_PATH}/question.jsonl', lines=True)
question_df['ground_truth'] = text_lines
question_df = question_df[question_df['ground_truth'] != 'TIE']

question_df
```

## Load model answers

We load answers from Alpaca-13B, GPT-3.5, GPT-4, and Vicuna-13B for the sampled dataset.


```python
answer_alpaca_df = pd.read_json(f'{VICUNA_DATASET_PATH}/answer/answer_alpaca-13b.jsonl', lines=True)
answer_gpt_35_df = pd.read_json(f'{VICUNA_DATASET_PATH}/answer/answer_gpt35.jsonl', lines=True)
answer_gpt_4_df = pd.read_json(f'{VICUNA_DATASET_PATH}/answer/answer_gpt-4.jsonl', lines=True)
answer_vicuna_df = pd.read_json(f'{VICUNA_DATASET_PATH}/answer/answer_vicuna-13b.jsonl', lines=True)

categories = question_df['category'].unique()

categories
```

## Sample questions

We randomly sample **one question per category** and store the IDs in `sampled_list_ids`. Repeat this **4** times (for 4 experiment configurations)


```python
sampled_list_ids = []

for i in range(4):
    current_list = []

    for category in categories:
        topic_questions = question_df[question_df['category'] == category]
        question_id = random.sample(topic_questions['question_id'].to_list(), 1)
        current_list.extend(question_id)
        
    sampled_list_ids.append(current_list)
    
sampled_list_ids
```

## Save sampled questions and answers

We save the sampled subset for all models in `../datasets/vicuna/sampled_data/cj_sampled`.

```python
SAVE_PATH = "../datasets/vicuna/sampled_data/cj_sampled"

for i in range(4):
    current_list_ids = sampled_list_ids[i]
    
    save_items(question_df, id_field='question_id', id_list=current_list_ids,
               save_path=f'{SAVE_PATH}/questions/{i}.jsonl')
    
    save_items(answer_alpaca_df, id_field='question_id', id_list=current_list_ids,
               save_path=f'{SAVE_PATH}/answer_alpaca/{i}.jsonl')
    save_items(answer_gpt_35_df, id_field='question_id', id_list=current_list_ids,
               save_path=f'{SAVE_PATH}/answer_gpt35/{i}.jsonl')
    save_items(answer_gpt_4_df, id_field='question_id', id_list=current_list_ids,
               save_path=f'{SAVE_PATH}/answer_gpt4/{i}.jsonl')
    save_items(answer_vicuna_df, id_field='question_id', id_list=current_list_ids,
               save_path=f'{SAVE_PATH}/answer_vicuna/{i}.jsonl')
```

# Merging vicuna dataset samples for CJ Experiments (Vanilla & MEC + BPC)

This notebook merges the sampled **questions** and **answers** for different models into unified datasets.  
We produce **vanilla** and **MEC+BPC** variants for both `vicuna` and `vicuna-gpt4` configurations.

**Final files:**
1. `vanilla-vicuna.jsonl`
2. `vanilla-vicuna-gpt4.jsonl`
3. `mec-bpc-vicuna.jsonl`
4. `mec-bpc-vicuna-gpt4.jsonl`

```python
# Paths
BASE_PATH = "../datasets/vicuna/sampled_data/cj_sampled"
MERGED_PATH = "../datasets/vicuna/sampled_data/cj_sampled_merged"
os.makedirs(MERGED_PATH, exist_ok=True)

# Number of repetitions for vanilla datasets
REPEAT_COUNT = 40
```

## Merge function

This merges `questions` with given model answers and repeats each row.

```python
def merge_dataset(q_path, save_path, **answers):
    """
    Merge questions and answers into one DataFrame,
    drop unused columns, rename 'text' to 'question_text',
    and repeat rows REPEAT_COUNT times.
    """
    q_df = pd.read_json(q_path, lines=True)
    merged_df = q_df.copy()

    for name, path in answers.items():
        df = pd.read_json(path, lines=True)
        merged_df = merged_df.merge(df, on="question_id", suffixes=("", f"_{name}"))

    # Drop unnecessary columns & rename text
    drop_cols = ['metadata', 'model_id', 'answer_id']
    
    merged_df = merged_df.drop(columns=[c for c in drop_cols if c in merged_df.columns], errors='ignore')
    if 'text' in merged_df.columns:
        merged_df = merged_df.rename(columns={'text': 'question_text'})

    # Repeat rows
    merged_df = pd.concat([merged_df] * REPEAT_COUNT, ignore_index=True)

    # Save
    merged_df.to_json(save_path, orient="records", lines=True)
    print(f"Saved merged dataset: {save_path} ({len(merged_df)} rows)")
    return merged_df
```

## Create Vanilla Datasets


```python
# Vanilla Vicuna
vanilla_vicuna = merge_dataset(
    save_path=f"{MERGED_PATH}/vanilla-vicuna.json",
    q_path=f"{BASE_PATH}/questions/0.jsonl",
    gpt35=f"{BASE_PATH}/answer_gpt35/0.jsonl",
    vicuna=f"{BASE_PATH}/answer_vicuna/0.jsonl"
)

# Vanilla Vicuna + GPT-4
vanilla_vicuna_gpt4 = merge_dataset(
    save_path=f"{MERGED_PATH}/vanilla-vicuna-gpt4.json",
    q_path=f"{BASE_PATH}/questions/1.jsonl",
    gpt35=f"{BASE_PATH}/answer_gpt35/1.jsonl",
    vicuna=f"{BASE_PATH}/answer_vicuna/1.jsonl"
)

vanilla_vicuna
```

## Create MEC+BPC Variants

We:
1. Repeat each row 6 times
2. Add `repeat_id` (1–6)
3. For repeats 4–6, swap `vicuna` and `gpt` answers
4. Drop `repeat_id` before saving

```python
def make_mec_bpc(df, vicuna_col, gpt_col, save_path):
    """
    Create MEC+BPC dataset by repeating rows ×6, swapping Vicuna/GPT answers for reversed repeats,
    and applying column drop + rename before repetition.
    """
    # Drop unnecessary columns & rename text
    drop_cols = ['metadata', 'model_id', 'answer_id']
    
    df = df.drop(columns=[c for c in drop_cols if c in df.columns], errors='ignore')
    if 'text' in df.columns:
        df = df.rename(columns={'text': 'question_text'})

    # Repeat ×6 for MEC+BPC structure
    mec_expanded = df.loc[df.index.repeat(6)].reset_index(drop=True)
    mec_expanded['repeat_id'] = np.tile(np.arange(1, 7), len(df))
    mec_expanded['is_reversed'] = mec_expanded['repeat_id'] > 3

    # Swap Vicuna and GPT answers for reversed repeats
    mask = mec_expanded['is_reversed']
    swapped = mec_expanded.loc[mask, [vicuna_col, gpt_col]].copy()
    mec_expanded.loc[mask, vicuna_col] = swapped[gpt_col]
    mec_expanded.loc[mask, gpt_col] = swapped[vicuna_col]

    # Drop helper columns
    mec_expanded.drop(columns=['repeat_id', 'is_reversed'], inplace=True)

    # Save
    mec_expanded.to_json(save_path, orient="records", lines=True)
    print(f"Saved MEC+BPC dataset: {save_path} ({len(mec_expanded)} rows)")
    return mec_expanded

# MEC+BPC Vicuna
mec_bpc_vicuna = make_mec_bpc(vanilla_vicuna, "text_vicuna", "text_gpt35",
             f"{MERGED_PATH}/mec-bpc-vicuna.json")

# MEC+BPC Vicuna + GPT-4
make_mec_bpc(vanilla_vicuna_gpt4, "text_vicuna", "text_gpt35",
             f"{MERGED_PATH}/mec-bpc-vicuna-gpt4.json")

mec_bpc_vicuna
```

# CS test dataset preparation

In this chapter, I create the datasets used for the experiment that evaluates the stability of results when introducing the **Calibrated Scorer** approach.  

The prepared datasets will serve as the input for running multiple experimental configurations, enabling a direct comparison between baseline and calibrated evaluation outcomes.

This chapter must be run after the script [cj_test.sh](../scripts/cj_test.sh) to use gathered results

```python
# Base inputs
DATA_DIR = Path("../datasets/vicuna/sampled_data/cj_sampled_merged")
RESULTS_DIR = Path("../gathered_data/cascaded-eval-results/final_results")
OUT_DIR = Path("../datasets/alpaca/cj-cs-test")

DATA_DIR
```

## Load ground truth for Calibrated Judge + Calibrated Scorer tests

I load the **ground-truth labels** from  
`../datasets/vicuna/ground-truth/cs_cj_ground_truth.txt`.  

These labels exclude possible ties (`TIE`), which are **accepted as a valid baseline outcome** for evaluation (VICUNA13B).

```python
with open('../datasets/vicuna/ground-truth/cs_cj_ground_truth.txt', 'r', encoding='utf-8') as f:
    lc_text_lines = f.read().splitlines()

print(f"Loaded {len(lc_text_lines)} ground-truth labels.")
```

## Load merged MEC+BPC datasets and attach ground truth

We read the two dataset variants:
- `mec-bpc-cj_sampled_merged.json` → MEC+BPC Vicuna GPT-3.5-turbo baseline
- `mec-bpc-cj_sampled_merged-gpt4.json` → MEC+BPC Vicuna GPT-4 baseline

```python
# GPT-4 baseline dataset
vicuna_mec_questions_gpt4 = pd.read_json(DATA_DIR / "mec-bpc-vicuna-gpt4.json")
vicuna_mec_questions_gpt4["ground_truth"] = vicuna_mec_questions_gpt4["question_id"].apply(lambda x: lc_text_lines[x - 1])
vicuna_mec_questions_gpt4 = vicuna_mec_questions_gpt4.set_index("question_id")

# GPT-3.5 baseline dataset
vicuna_mec_questions = pd.read_json(DATA_DIR / "mec-bpc-vicuna.json")
vicuna_mec_questions["ground_truth"] = vicuna_mec_questions["question_id"].apply(lambda x: lc_text_lines[x - 1])
vicuna_mec_questions = vicuna_mec_questions.set_index("question_id")

vicuna_mec_questions_gpt4
```

## Load per-run CascadedEval results

For each run (1–40), we:
- Load **Auto-J** and **JudgeLM** JSON outputs.
- Load both GPT-4 and GPT-3.5 baseline versions.
- Attach the `score` column to the corresponding `vicuna_mec_questions` DataFrame.

```python
def load_json(path: Path):
    with open(path, "r", encoding="utf-8") as f:
        return json.loads(f.read().strip())
```

```python
final_list = []

for run_idx in range(1, 41):
    # File paths for this run
    auto_j_gpt4_path    = RESULTS_DIR / "vicuna-mec-gpt4" / f"{run_idx}-auto-j-vicuna-mec-gpt4-final.json"
    auto_j_path         = RESULTS_DIR / "vicuna-mec"      / f"{run_idx}-auto-j-vicuna-mec-final.json"
    judgelm_gpt4_path   = RESULTS_DIR / "vicuna-mec-gpt4" / f"{run_idx}-judgelm-vicuna-mec-gpt4-final.json"
    judgelm_path        = RESULTS_DIR / "vicuna-mec"      / f"{run_idx}-judgelm-vicuna-mec-final.json"

    # Load raw judgement lists
    auto_j_results_gpt4  = load_json(auto_j_gpt4_path)
    auto_j_results       = load_json(auto_j_path)
    judgelm_results_gpt4 = load_json(judgelm_gpt4_path)
    judgelm_results      = load_json(judgelm_path)

    # Attach scores to base question data
    auto_j = vicuna_mec_questions.copy()
    auto_j["score"] = auto_j_results

    auto_j_gpt4 = vicuna_mec_questions_gpt4.copy()
    auto_j_gpt4["score"] = auto_j_results_gpt4

    judgelm = vicuna_mec_questions.copy()
    judgelm["score"] = judgelm_results

    judgelm_gpt4 = vicuna_mec_questions_gpt4.copy()
    judgelm_gpt4["score"] = judgelm_results_gpt4

    final_list.append({
        "auto-j-gpt4": auto_j_gpt4,
        "auto-j": auto_j,
        "judgelm": judgelm,
        "judgelm-gpt4": judgelm_gpt4
    })
    
final_list
```

## Reverse rows where needed

Some experiments are **reversed** (model responses swapped).  
I fix that by applying the helper `reverse_row` to every DataFrame.

```python
for key in final_list[0].keys():
    for res in final_list:
        res[key] = res[key].apply(reverse_row, axis=1)

final_list
```

## Extract categorical results

We convert each row into a categorical label (`CHATGPT`, `VICUNA13B`)  
using `extract_result()` which applies the LC ground truth mapping and tie rules.


```python
for key in final_list[0].keys():
    for res in final_list:
        res[key] = res[key].apply(lambda row: extract_result(row, lc_text_lines, tie_label="VICUNA13B"), axis=1)
        
final_list
```

## Aggregate to majority vote per question

I group by `question_id` within each run and:
- Keep the first value for metadata columns
- Use majority voting for the `result` column
- Compute `result_hit` — whether `result` matches `ground_truth`


```python
for current_eval_result in final_list:
    for key, df in current_eval_result.items():
        value = df.groupby("question_id").agg(
            {**{col: "first" for col in df.columns if col not in ["question_id", "result"]},
             "result": lambda x: x.value_counts().idxmax()}
        )
        value["result_hit"] = value["result"] == value["ground_truth"]
        current_eval_result[key] = value

final_list
```

## Compute preference scores

We add a `preference` column:  
`1` if `result == "CHATGPT"`, otherwise `0`.


```python
for key in final_list[0].keys():
    for res in final_list:
        res[key]["preference"] = (res[key]["result"] == "CHATGPT").astype(int)
        
final_list
```

## Normalize schema

I rename columns to a unified format and drop helper columns (`category`, `is_reversed`, `score`, `result_hit`).
Also set consistent `baseline_name` and `model_name` values.


```python
for current_eval_result in final_list:
    for key, df in current_eval_result.items():
        df = df.rename(columns={
            "question_text": "instruction",
            "text_gpt": "baseline_response",
            "text_vicuna": "model_response",
        })
        drop_cols = [c for c in ["category", "is_reversed", "score", "result_hit"] if c in df.columns]
        df = df.drop(columns=drop_cols, errors="ignore")

        df["baseline_name"] = "vicuna13b"
        df["model_name"] = "gpt-3.5-turbo"

        current_eval_result[key] = df

final_list
```

## Concatenate across runs

Icombine all runs into final large DataFrames for each model/configuration pair.


```python
auto_j = pd.concat([item["auto-j"] for item in final_list]).reset_index()
auto_j_gpt4 = pd.concat([item["auto-j-gpt4"] for item in final_list]).reset_index()

judgelm = pd.concat([item["judgelm"] for item in final_list]).reset_index()
judgelm_gpt4 = pd.concat([item["judgelm-gpt4"] for item in final_list]).reset_index()

auto_j
```

## Save final artifacts

I save the four processed datasets into `cj_cs_introduction_results` as JSON arrays.


```python
auto_j.to_json(OUT_DIR / "alpaca-auto-j.json", orient="records")
auto_j_gpt4.to_json(OUT_DIR / "alpaca-auto-j-gpt4.json", orient="records")
judgelm.to_json(OUT_DIR / "alpaca-judgelm.json", orient="records")
judgelm_gpt4.to_json(OUT_DIR / "alpaca-judgelm-gpt4.json", orient="records")

auto_j.head(2), auto_j_gpt4.head(2), judgelm.head(2), judgelm_gpt4.head(2)
```
