# Library

The same functions as in [experimental_result_analysis.ipynb](./experimental_result_analysis.ipynb)

ID's extraction of less confident results

```python
def extract_non_confident_indices(result_df, ratio=0.5):
    relia_scores = result_df['entropy'].to_numpy()

    sorted_indices = np.argsort(-relia_scores)
    top_half_indices = sorted_indices[:int(len(sorted_indices) * ratio)]

    top_df_indices = result_df.index[top_half_indices]

    return set(result_df.index) - set(top_df_indices)
```

Extract the ground truth and result

```python
def extract_result(row, ground_truth, win_label="CHATGPT", lose_label="VICUNA13B", tie_label="TIE"):
    first_score, second_score = row['score']

    row['result'] = win_label if round(first_score, 2) > round(second_score, 2) else lose_label if round(first_score,
                                                                                                         2) < round(
        second_score, 2) else tie_label
    row['ground_truth'] = ground_truth[row.name - 1]
    row['result_hit'] = row['result'] == row['ground_truth']

    return row
```

Reverse score

```python
def reverse_row(row):
    first_score, second_score = row['score']
    is_reversed = row['is_reversed']

    if is_reversed:
        row['score'] = [second_score, first_score]
        row['text_gpt'], row['text_vicuna'] = row['text_vicuna'], row['text_gpt']

    return row
```

# Merging vicuna dataset samples for CJ Experiments (Vanilla & MEC + BPC)

This notebook merges the sampled **questions** and **answers** for different models into unified datasets.  
We produce **vanilla** and **MEC+BPC** variants for both `vicuna` and `vicuna-gpt4` configurations.

**Final files:**

```python
# Paths
BASE_PATH = "../datasets/vicuna/sampled_data/cj_sampled"
MERGED_PATH = "../datasets/vicuna/sampled_data/cj_sampled_merged"
os.makedirs(MERGED_PATH, exist_ok=True)

# Number of repetitions for vanilla datasets
REPEAT_COUNT = 40
```

## Merge function

This merges `questions` with given model answers and repeats each row.

```python
def merge_dataset(q_path, save_path, **answers):
    """
    Merge questions and answers into one DataFrame,
    drop unused columns, rename 'text' to 'question_text',
    and repeat rows REPEAT_COUNT times.
    """
    q_df = pd.read_json(q_path, lines=True)
    merged_df = q_df.copy()

    for name, path in answers.items():
        df = pd.read_json(path, lines=True)
        merged_df = merged_df.merge(df, on="question_id", suffixes=("", f"_{name}"))

    # Drop unnecessary columns & rename text
    drop_cols = ['metadata', 'model_id', 'answer_id']
    
    merged_df = merged_df.drop(columns=[c for c in drop_cols if c in merged_df.columns], errors='ignore')
    if 'text' in merged_df.columns:
        merged_df = merged_df.rename(columns={'text': 'question_text'})

    # Repeat rows
    merged_df = pd.concat([merged_df] * REPEAT_COUNT, ignore_index=True)

    # Save
    merged_df.to_json(save_path, orient="records", lines=True)
    print(f"Saved merged dataset: {save_path} ({len(merged_df)} rows)")
    return merged_df
```

## Create Vanilla Datasets


```python
# Vanilla Vicuna
vanilla_vicuna = merge_dataset(
    save_path=f"{MERGED_PATH}/vanilla-vicuna.json",
    q_path=f"{BASE_PATH}/questions/0.jsonl",
    gpt35=f"{BASE_PATH}/answer_gpt35/0.jsonl",
    vicuna=f"{BASE_PATH}/answer_vicuna/0.jsonl"
)

# Vanilla Vicuna + GPT-4
vanilla_vicuna_gpt4 = merge_dataset(
    save_path=f"{MERGED_PATH}/vanilla-vicuna-gpt4.json",
    q_path=f"{BASE_PATH}/questions/1.jsonl",
    gpt35=f"{BASE_PATH}/answer_gpt35/1.jsonl",
    vicuna=f"{BASE_PATH}/answer_vicuna/1.jsonl"
)

vanilla_vicuna
```

## Create MEC+BPC Variants

We:
1. Repeat each row 6 times
2. Add `repeat_id` (1–6)
3. For repeats 4–6, swap `vicuna` and `gpt` answers
4. Drop `repeat_id` before saving

```python
def make_mec_bpc(df, vicuna_col, gpt_col, save_path):
    """
    Create MEC+BPC dataset by repeating rows ×6, swapping Vicuna/GPT answers for reversed repeats,
    and applying column drop + rename before repetition.
    """
    # Drop unnecessary columns & rename text
    drop_cols = ['metadata', 'model_id', 'answer_id']
    
    df = df.drop(columns=[c for c in drop_cols if c in df.columns], errors='ignore')
    if 'text' in df.columns:
        df = df.rename(columns={'text': 'question_text'})

    # Repeat ×6 for MEC+BPC structure
    mec_expanded = df.loc[df.index.repeat(6)].reset_index(drop=True)
    mec_expanded['repeat_id'] = np.tile(np.arange(1, 7), len(df))
    mec_expanded['is_reversed'] = mec_expanded['repeat_id'] > 3

    # Swap Vicuna and GPT answers for reversed repeats
    mask = mec_expanded['is_reversed']
    swapped = mec_expanded.loc[mask, [vicuna_col, gpt_col]].copy()
    mec_expanded.loc[mask, vicuna_col] = swapped[gpt_col]
    mec_expanded.loc[mask, gpt_col] = swapped[vicuna_col]

    # Drop helper columns
    mec_expanded.drop(columns=['repeat_id', 'is_reversed'], inplace=True)

    # Save
    mec_expanded.to_json(save_path, orient="records", lines=True)
    print(f"Saved MEC+BPC dataset: {save_path} ({len(mec_expanded)} rows)")
    return mec_expanded

# MEC+BPC Vicuna
mec_bpc_vicuna = make_mec_bpc(vanilla_vicuna, "text_vicuna", "text_gpt35",
             f"{MERGED_PATH}/mec-bpc-vicuna.json")

# MEC+BPC Vicuna + GPT-4
make_mec_bpc(vanilla_vicuna_gpt4, "text_vicuna", "text_gpt35",
             f"{MERGED_PATH}/mec-bpc-vicuna-gpt4.json")

mec_bpc_vicuna
```

For each run (1–40), we:
- Load **Auto-J** and **JudgeLM** JSON outputs.
- Load both GPT-4 and GPT-3.5 baseline versions.
- Attach the `score` column to the corresponding `vicuna_mec_questions` DataFrame.

```python
def load_json(path: Path):
    with open(path, "r", encoding="utf-8") as f:
        return json.loads(f.read().strip())
```

```python
final_list = []

for run_idx in range(1, 41):
    # File paths for this run
    auto_j_gpt4_path    = RESULTS_DIR / "vicuna-mec-gpt4" / f"{run_idx}-auto-j-vicuna-mec-gpt4-final.json"
    auto_j_path         = RESULTS_DIR / "vicuna-mec"      / f"{run_idx}-auto-j-vicuna-mec-final.json"
    judgelm_gpt4_path   = RESULTS_DIR / "vicuna-mec-gpt4" / f"{run_idx}-judgelm-vicuna-mec-gpt4-final.json"
    judgelm_path        = RESULTS_DIR / "vicuna-mec"      / f"{run_idx}-judgelm-vicuna-mec-final.json"

    # Load raw judgement lists
    auto_j_results_gpt4  = load_json(auto_j_gpt4_path)
    auto_j_results       = load_json(auto_j_path)
    judgelm_results_gpt4 = load_json(judgelm_gpt4_path)
    judgelm_results      = load_json(judgelm_path)

    # Attach scores to base question data
    auto_j = vicuna_mec_questions.copy()
    auto_j["score"] = auto_j_results

    auto_j_gpt4 = vicuna_mec_questions_gpt4.copy()
    auto_j_gpt4["score"] = auto_j_results_gpt4

    judgelm = vicuna_mec_questions.copy()
    judgelm["score"] = judgelm_results

    judgelm_gpt4 = vicuna_mec_questions_gpt4.copy()
    judgelm_gpt4["score"] = judgelm_results_gpt4

    final_list.append({
        "auto-j-gpt4": auto_j_gpt4,
        "auto-j": auto_j,
        "judgelm": judgelm,
        "judgelm-gpt4": judgelm_gpt4
    })
    
final_list
```

## Reverse rows where needed

Some experiments are **reversed** (model responses swapped).  
I fix that by applying the helper `reverse_row` to every DataFrame.

```python
for key in final_list[0].keys():
    for res in final_list:
        res[key] = res[key].apply(reverse_row, axis=1)

final_list
```

## Extract categorical results

We convert each row into a categorical label (`CHATGPT`, `VICUNA13B`)  
using `extract_result()` which applies the LC ground truth mapping and tie rules.


```python
for key in final_list[0].keys():
    for res in final_list:
        res[key] = res[key].apply(lambda row: extract_result(row, lc_text_lines, tie_label="VICUNA13B"), axis=1)
        
final_list
```

## Aggregate to majority vote per question

I group by `question_id` within each run and:
- Keep the first value for metadata columns
- Use majority voting for the `result` column
- Compute `result_hit` — whether `result` matches `ground_truth`


```python
for current_eval_result in final_list:
    for key, df in current_eval_result.items():
        value = df.groupby("question_id").agg(
            {**{col: "first" for col in df.columns if col not in ["question_id", "result"]},
             "result": lambda x: x.value_counts().idxmax()}
        )
        value["result_hit"] = value["result"] == value["ground_truth"]
        current_eval_result[key] = value

final_list
```

## Compute preference scores

We add a `preference` column:  
`1` if `result == "CHATGPT"`, otherwise `0`.


```python
for key in final_list[0].keys():
    for res in final_list:
        res[key]["preference"] = (res[key]["result"] == "CHATGPT").astype(int)
        
final_list
```

## Normalize schema

I rename columns to a unified format and drop helper columns (`category`, `is_reversed`, `score`, `result_hit`).
Also set consistent `baseline_name` and `model_name` values.


```python
for current_eval_result in final_list:
    for key, df in current_eval_result.items():
        df = df.rename(columns={
            "question_text": "instruction",
            "text_gpt": "baseline_response",
            "text_vicuna": "model_response",
        })
        drop_cols = [c for c in ["category", "is_reversed", "score", "result_hit"] if c in df.columns]
        df = df.drop(columns=drop_cols, errors="ignore")

        df["baseline_name"] = "vicuna13b"
        df["model_name"] = "gpt-3.5-turbo"

        current_eval_result[key] = df

final_list
```

## Concatenate across runs

Icombine all runs into final large DataFrames for each model/configuration pair.


```python
auto_j = pd.concat([item["auto-j"] for item in final_list]).reset_index()
auto_j_gpt4 = pd.concat([item["auto-j-gpt4"] for item in final_list]).reset_index()

judgelm = pd.concat([item["judgelm"] for item in final_list]).reset_index()
judgelm_gpt4 = pd.concat([item["judgelm-gpt4"] for item in final_list]).reset_index()

auto_j
```

# CS verbosity experiment dataset preparation

**Methodology**  
- Use AlpacaEval instructions from 5 datasets.  
- For each dataset, **randomly sample 10 instructions** (fixed seed) for consistency.  
- Compare each selected model **vs the baseline** (`gpt-4-1106-preview`) across three verbosity settings (as provided in the source outputs).  
- Build balanced judgment pairs by **mirroring** (swap answer/model positions) and **replicating** each orientation **3×** to reduce position effects.  


```python
# ---- Configuration ----
BASELINE_NAME = "gpt4_1106_preview"  # baseline id used in the local AlpacaEval dumps
BASE_ALPACA_PATH = "../datasets/alpaca/framework_datasets"

# Only include these four models in the analysis (including the baseline). Adjust names to match folder names.
SELECTED_MODELS = [
    "gpt4_1106_preview",                # baseline
    "mixtral_8x7b_instruct_v0_1",
    "claude_2_1",
    "gpt3_5_turbo_1106",
]

# Fixed seed for reproducibility
RANDOM_STATE = 42
```

## Discover available model outputs

I list folders under `BASE_ALPACA_PATH` and **filter** them to the four target models to ensure we only include the intended systems.

```python
all_dirs = [name for name in os.listdir(BASE_ALPACA_PATH) if os.path.isdir(os.path.join(BASE_ALPACA_PATH, name))]
all_dirs.sort()

if BASELINE_NAME not in all_dirs:
    raise FileNotFoundError(f"Baseline '{BASELINE_NAME}' not found among available dirs: {dirs}")
all_dirs
```

## Load baseline model outputs

We load the baseline responses for later pairing.

```python
baseline_path = f"{BASE_ALPACA_PATH}/{BASELINE_NAME}/model_outputs.json"
baseline_outputs = pd.read_json(baseline_path)

baseline_outputs
```

## Sample 10 instructions per dataset

We sample on the **first non-baseline** model's outputs to pick the instruction set: **10 per dataset**, fixed `random_state`.

```python
# Choose a reference model (first non-baseline) to define the instruction sample
ref_models = [d for d in all_dirs if d != BASELINE_NAME]

print(ref_models)
if not ref_models:
    raise RuntimeError("No non-baseline models available to sample instructions from.")
ref_model = ref_models[0]

model_outputs = pd.read_json(f"{BASE_ALPACA_PATH}/{ref_model}/model_outputs.json")
# Expect columns like: ['dataset', 'instruction', 'output', ...]
instructions = (
    model_outputs
    .groupby('dataset', group_keys=False)
    .sample(n=10, random_state=RANDOM_STATE)['instruction']
    .unique()
)

len(instructions)
```

## Build pairwise judgments against the baseline

For each selected model and each sampled instruction, we form a pair `(answer_1 vs answer_2)` as `(model, baseline)`.  
This enables relative evaluation (win rates) following the procedures in the paper.


```python
judgement_objects = []

for name in all_dirs:
    current_outputs = pd.read_json(f"{BASE_ALPACA_PATH}/{name}/model_outputs.json")
    
    # Fast index by instruction for lookups
    idx_current = current_outputs.set_index('instruction')
    idx_baseline = baseline_outputs.set_index('instruction')
    
    for instruction in instructions:
        if instruction not in idx_current.index or instruction not in idx_baseline.index:
            # Skip missing instructions (robustness if dumps are not perfectly aligned)
            continue
            
        judgement_objects.append({
            'instruction': instruction,
            'answer_1': idx_current.loc[instruction, 'output'],
            'answer_2': idx_baseline.loc[instruction, 'output'],
            'model_1': name,
            'model_2': BASELINE_NAME
        })

len(judgement_objects)
```

## Create the base judgment DataFrame


```python
judgement_df = pd.DataFrame(judgement_objects).reset_index(drop=True)
judgement_df
```

## MEC+BPC implementation

- For each row, **replicate 3×** in the original order.  
- Create a **swapped** version (`answer_1 ↔ answer_2`, `model_1 ↔ model_2`) and **replicate 3×** as well.  
This yields a balanced, position-robust dataset.

```python
rows = []

for _, row in judgement_df.iterrows():
    for _ in range(3):
        rows.append(row.copy())

    swapped_row = row.copy()
    swapped_row['answer_1'], swapped_row['answer_2'] = row['answer_2'], row['answer_1']
    swapped_row['model_1'], swapped_row['model_2'] = row['model_2'], row['model_1']
    for _ in range(3):
        rows.append(swapped_row.copy())

df_judgement_expanded = pd.DataFrame(rows).reset_index(drop=True)
df_judgement_expanded
```

## Save the expanded dataset

The final file is written to `../gathered_data/verbosity_datasets/judgement_expanded.json` (relative to the notebook).

```python
out_dir = Path("../datasets/alpaca/cj-cs-test")
out_dir.mkdir(parents=True, exist_ok=True)

out_path = out_dir / "judgement_expanded.json"
df_judgement_expanded.to_json(out_path, orient='records')
out_path.as_posix()
```

# CS + CJ dataset preparation

This section prepares a **dataset** for combined CS + CJ evaluation, starting from cascaded evaluation results and applying filtering, renaming, preference labeling, and balancing.

```python
DROP_LIST = [
    'alpaca-7b_verbose', 'alpaca-7b', 'alpaca-7b_concise',
    'gpt4_0613_verbose', 'gpt4_0613_concise', 'gpt4_0613'
]
```

## Load cascaded evaluation results for verbosity test

```python
final_combination_df = pd.read_json("../gathered_data/cascaded-eval-results/final_results/verbosity.json")
final_combination_df
```

## Swap model_1 and model_2 for positional symmetry, and flip score ordering

```python
swapped_df = final_combination_df.copy()

for i in range(0, len(swapped_df), 6):
    swap_slice = swapped_df.iloc[i + 3:i + 6].copy()

    swapped_df.iloc[i + 3:i + 6, swapped_df.columns.get_loc('model_1')] = swap_slice['model_2'].values
    swapped_df.iloc[i + 3:i + 6, swapped_df.columns.get_loc('model_2')] = swap_slice['model_1'].values
    swapped_df.iloc[i + 3:i + 6, swapped_df.columns.get_loc('score')] = (
        swap_slice['score'].apply(lambda x: [x[1], x[0]]).values
    )

# Recompute result column after swapping
swapped_df['result'] = swapped_df.apply(
    lambda row: row['model_1'] if row['score'][0] >= row['score'][1] else row['model_2'],
    axis=1
)

swapped_df
```

## Aggregate results by instruction and model pair, taking the majority result

```python
agg_dict = {
    col: (lambda x: x.value_counts().idxmax()) if col == 'result' else 'first'
    for col in swapped_df.columns
}

grouped_final_combination = (
    swapped_df
    .groupby(['instruction', 'model_1', 'model_2'])
    .agg(agg_dict)
    .drop(['instruction', 'score', 'entropy', 'model_1', 'model_2'], axis=1)
)

grouped_final_combination
```

## Drop unwanted models listed in `DROP_LIST`

```python
dropped_df = grouped_final_combination[
    ~grouped_final_combination.index.get_level_values('model_1').isin(DROP_LIST) &
    ~grouped_final_combination.index.get_level_values('model_2').isin(DROP_LIST)
]
dropped_df
```

## Rename columns for consistency in downstream tasks

```python
renamed_df = (
    dropped_df
    .reset_index()
    .rename(columns={
        'model_1': 'model_name',
        'model_2': 'baseline_name',
        'answer_1': 'model_response',
        'answer_2': 'baseline_response'
    })
)
renamed_df
```

## Define helper functions for generating preferences

```python
BASELINE_NAME = 'gpt4_1106_preview'

def alternating_zeros_ones():
    while True:
        yield 0
        yield 1

gen = alternating_zeros_ones()

def extract_preferences(row):
    if row['model_name'] != BASELINE_NAME:
        return 1 if row['result'] != BASELINE_NAME else 0
    else:
        return next(gen)
```

## Assign preference labels to each row

```python
final_verb_df = renamed_df.copy()
final_verb_df['preference'] = final_verb_df.apply(extract_preferences, axis=1)
final_verb_df
```

## Balance preferences for each (baseline, model) pair

```python
balanced_rows = []

for (baseline, model), group in final_verb_df.groupby(['baseline_name', 'model_name']):
    group_0 = group[group['preference'] == 0]
    group_1 = group[group['preference'] == 1]
    n = min(len(group_0), len(group_1))

    if n == 0:
        continue

    more_zeros = len(group_0) > len(group_1)
    sampled_0 = group_0.sample(n=n if not more_zeros else min(4 * n, len(group_0)), random_state=42)
    sampled_1 = group_1.sample(n=n if more_zeros else min(4 * n, len(group_1)), random_state=42)

    balanced_rows.append(pd.concat([sampled_0, sampled_1]))

balanced_df = pd.concat(balanced_rows).reset_index(drop=True)
balanced_df
```

## Save the balanced dataset to JSON

```python
balanced_df.to_json('../datasets/alpaca/cj-cs-test/balanced.json', orient='records')
```
